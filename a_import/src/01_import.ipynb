```python

import json
from pathlib import Path
import pandas as pd
```

```python
ROOT = Path.cwd().resolve()
if ROOT.name == 'src':
    ROOT = ROOT.parent.parent
SEED = 63
```

```python
config = {
    'seed': SEED, 
    'nx': 121, 'ny': 257, 'source_y': 129, 
    'packets_per_step': 80,
    'diffusivity': 0.22, 
    'wind_bias': 0.30, 'wind_blocks': [1, 8, 32],
    'n_trials': 64, 'n_steps': 1024, 
    'burn_in': 121, 'baseline_steps': 256,
    'receptor_x': 48, 'receptor_offsets': [-12, 0, 12], 'receptor_halfwidth': 2,
    'exposure_threshold': 4.0, 'profile_distances': [12, 30, 48],
    'frame_interval': 8, 'demo_frames': 180, 
    'bootstrap_draws': 63000,
    'weather_demo': {'gentle_drift': 0.15, 'x_min': -80, 'x_max': 143, 'y_halfwidth': 80}
}

```

```python
for task in ['a_import', 'b_simulate', 'c_analyze']:
    for folder in ['input', 'output', 'src']:
        (ROOT / task / folder).mkdir(parents=True, exist_ok=True)
(ROOT / 'a_import/output/config.json').write_text(json.dumps(config, indent=2))

for task, name, target in [
    ('b_simulate', 'upstream', '../../a_import/output'),
    ('c_analyze', 'upstream', '../../b_simulate/output')
]:
    link = ROOT / task / 'input' / name
    link.unlink(missing_ok=True)
    link.symlink_to(target, target_is_directory=True)
pd.DataFrame(config.items(), columns=['parameter', 'value'])
```
